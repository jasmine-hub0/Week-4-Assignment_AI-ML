```python
def clean_transactions(amounts, min_threshold=10.0):
    valid_amounts = list(filter(
        lambda amount: amount > 0 and amount >= min_threshold,
        amounts
    ))
    return valid_amounts
```

```python
def apply_service_fee(amounts, fee_rate=0.02):
    processed_amounts = list(map(
        lambda amount: round(amount - (amount * fee_rate), 2),
        amounts
    ))
    return processed_amounts
```

```python
def calculate_nested_total(data):
    # Base case: an empty list
    if data == []:
        return 0.0

    # Base case: a single number
    if isinstance(data, (int, float)):
        return data

    # Recursive step: calculate head and tail
    head = data[0]
    tail = data[1:]

    return calculate_nested_total(head) + calculate_nested_total(tail)
```

```python
# Test the transaction cleaning function
transactions = [100, 5, 250, -20, 15, 0, 8, 500]

cleaned = clean_transactions(transactions)
print("Original Transactions:", transactions)
print("Cleaned Transactions:", cleaned)
```

```python
# Test the service fee function
processed = apply_service_fee(cleaned)
print("After Service Fee:", processed)
```

```python
# Test the recursive total function
nested_transactions = [100, [50, 25], [[10, 5], 40]]
total = calculate_nested_total(nested_transactions)

print("Nested Transactions:", nested_transactions)
print("Total Transferred Amount:", total)
```

```python
# Test empty list and single number
print("Empty List Total:", calculate_nested_total([]))
print("Single Transaction Total:", calculate_nested_total(150.0))
```
